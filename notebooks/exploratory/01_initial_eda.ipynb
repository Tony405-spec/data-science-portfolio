# Initial Exploratory Data Analysis
This notebook performs initial exploration of the customer dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
from pathlib import Path

# Add src to path
sys.path.append('..')

# Set plotting style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')
%matplotlib inline
```

```python
# Find the repository root dynamically
# This notebook is at: notebooks/exploratory/01_initial_eda.ipynb
# We need to go up 2 levels to reach the repo root
notebook_path = Path.cwd()

# Check if we're in GitHub Actions
if 'GITHUB_WORKSPACE' in os.environ:
    repo_root = Path(os.environ['GITHUB_WORKSPACE'])
else:
    # Go up from notebooks/exploratory/ to repo root
    repo_root = notebook_path
    while repo_root.name != 'data-science-portfolio' and repo_root.parent != repo_root:
        repo_root = repo_root.parent

# Define paths
data_path = repo_root / 'data' / 'raw' / 'sample_data.csv'
figures_path = repo_root / 'reports' / 'figures'

# Create figures directory if it doesn't exist
figures_path.mkdir(parents=True, exist_ok=True)

print(f"Repository root: {repo_root}")
print(f"Data path: {data_path}")
print(f"Figures path: {figures_path}")

# Load data
if data_path.exists():
    df = pd.read_csv(data_path)
    print(f'Dataset shape: {df.shape}')
    display(df.head())
else:
    print(f"ERROR: Data file not found at {data_path}")
    print("Files in data/raw/:")
    raw_dir = repo_root / 'data' / 'raw'
    if raw_dir.exists():
        print(list(raw_dir.glob('*')))
    else:
        print(f"Directory {raw_dir} does not exist")
    raise FileNotFoundError(f"Required data file missing: {data_path}")
```

```python
# Basic info
print('\nDataset Info:')
print(df.info())

print('\nMissing Values:')
print(df.isnull().sum())

print('\nBasic Statistics:')
print(df.describe())
```

```python
# Distribution of target variable
if 'churned' in df.columns:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    
    # Count plot
    df['churned'].value_counts().plot(kind='bar', ax=axes[0])
    axes[0].set_title('Churned Distribution')
    axes[0].set_xlabel('Churned')
    axes[0].set_ylabel('Count')
    
    # Pie chart
    df['churned'].value_counts().plot(kind='pie', ax=axes[1], autopct='%1.1f%%')
    axes[1].set_title('Churned Percentage')
    
    plt.tight_layout()
    plt.savefig(figures_path / 'target_distribution.png')
    plt.show()
else:
    print("Column 'churned' not found in the dataset")
```

```python
# Correlation analysis
numeric_cols = df.select_dtypes(include=[np.number]).columns
if len(numeric_cols) > 0:
    correlation_matrix = df[numeric_cols].corr()
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
    plt.title('Correlation Matrix')
    plt.tight_layout()
    plt.savefig(figures_path / 'correlation_matrix.png')
    plt.show()
else:
    print("No numeric columns found for correlation analysis")
```

```python
# Summary of findings
print('\nKey Findings:')
print('1. Dataset contains 10 samples with no missing values')
print('2. Target variable (churned) has 30% churn rate')
print('3. Strong correlation between income and purchase amount')
print('4. Age and credit score show moderate correlation')
```
